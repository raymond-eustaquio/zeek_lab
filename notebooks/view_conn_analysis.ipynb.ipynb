```python
import pandas as pd
from io import StringIO

def load_conn_log(path):
    with open(path, "r") as f:
        lines = f.readlines()

    header = None
    for line in lines:
        if line.startswith("#fields"):
            header = line.strip().split()[1:]
            break

    data = [l for l in lines if not l.startswith("#")]

    return pd.read_csv(
        StringIO("".join(data)),
        sep="\t",
        names=header,
        engine="python"
    )

df = load_conn_log("logs/logs_reval-cap/conn.log")
df.head()
```

```python
hist = df[df["history"].notna()].copy()
hist.head()
```

```python
from collections import Counter

all_codes = Counter("".join(hist["history"].tolist()))
df_hist_freq = pd.DataFrame(all_codes.items(), columns=["Code", "Count"]).sort_values("Count", ascending=False)
df_hist_freq
```

```python
import pandas as pd
from collections import Counter

# Extract only rows that contain history codes
hist = df[df["history"].notna()].copy()

# Count individual history characters across all rows
all_codes = Counter("".join(hist["history"].tolist()))

# Convert to DataFrame for clean Jupyter display
df_hist_freq = (
    pd.DataFrame(all_codes.items(), columns=["Code", "Count"])
      .sort_values("Count", ascending=False)
      .reset_index(drop=True)
)

df_hist_freq
```
